### Task1

```python
import pandas as pd
marks_list = [78, 85, 90, 66, 72]
marks = pd.Series(marks_list)

print(f"Series value = {marks.values}, Index: {marks.index} and data type:{marks.dtype}")

# access elements
print(f"First element: {marks[0]} and last two elements are: {marks[3:].values}")
```

### Task2

```python
print(f"Original marks:{marks.values}")
marks+=5
print(f"Addition of 5 grace marks would result: {marks.values}")
marks-=2
print(f"Subtraction of 2 marks would result: {marks.values}")
marks*=1.05
print(f"Multiplication of 1.05 would result: {marks.values}")
marks/=2
print(f"Division of 2 would result: {marks.values}")
```

### Task3

```python
marks_list = [78, 85, 90, 66, 72]
marks = pd.Series(marks_list)
print(f"Max for the series is: {marks.max()}")
print(f"Min for the series is: {marks.min()}")
print(f"Sum for the series is: {marks.sum()}")
print(f"Mean for the series is: {marks.mean()}")

# lambda function
check_marks = lambda marks: True if marks>=70 else False
results = pd.Series(map(check_marks, marks))
print("Students Passed:\n",results)
print(f"No. of student passed: {results.sum()}")
```

### Task4

```python
students_data = {"Name": ['Amit', 'Neha', 'Rahul', 'Sneha', 'Pooja'], 
                 "Marks":[78, 85, 90, 66, 72],
                 "Subjects":[ 'Math', 'Math', 'Science', 'Science', 'Math']
                }
students = pd.DataFrame(students_data)
print("First three rows: ")
print(students.head(3))
print("Last two rows: ")
print(students.tail(2))
print(f"Dataframe shape is: {students.shape} and columns name are: {students.columns}")
```

### Task5

```python
print("INFO:")
print(students.info())
print("\n DESCRIBE:")
print(students.describe())
print("\n Head:")
print(students.head())
print("\n Tail:")
print(students.tail())

# sorting by marks in desending
print("\n MARKS IN DESENDING ORDER: ")
students.sort_values('Marks', ascending=False, inplace=True)
print(students)

print("\n BACK TO NORMAL DATA")
students.sort_index(inplace=True)
print(students)
```

### Task6

```python
print("1. Students with marks > 75 are: \n")
print(students[students['Marks'] > 75])

print("\n 2. Students with maths sub: \n")
print(students[students['Subjects'] == "Math"])

print("\n 3. Students scored more then avg: \n")
mean_marks = students['Marks'].mean()
print(students[students['Marks'] > mean_marks])

print("\n 4. Students who failed: \n")
print(students[students['Marks'] < 70])
```

### Task7

```python
print('Avg marks per subject will be: \n')
print(students[['Marks','Subjects']].groupby('Subjects').mean())

print('\n Number of students per subject: \n')
print(students['Subjects'].value_counts())

print('\n Max marks per subjects are: \n')
print(students[['Marks','Subjects']].groupby('Subjects').max())
```

### Task9

```python
sales_data = {
    'Day': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Revenue': [1200, 1500, 900, 2000, 1800]
}
sales = pd.DataFrame(sales_data)

print(f"Total revenue: {sales['Revenue'].sum()}")
print(f"Avg revenue: {sales['Revenue'].mean()} \n")

print(f"Day with heights revenue is: \n")
print(sales.sort_values('Revenue', ascending=False).head(1))

print('\nDays when revenue would be greater the avg: \n')
avg = sales['Revenue'].mean()
print(sales[sales['Revenue'] > avg])

sales.plot(x='Day', y='Revenue', title='Revenue vs Day')
```
