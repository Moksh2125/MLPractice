```python
A = np.array([1,2,3,4])
B = np.array([10,20,30,40])

print(A + B)
print(A - B)
print(A * B)
print(A / B)
print(A**2)

#extra
print(np.add(A,B), np.subtract(A,B))
```

## Task3

```python
values = np.array([2,4,6,8,10])

print(f" the square root is: {np.sqrt(values)}")
print(f" the Exponential is: {np.exp(values)}")
print(f" the Log is: {np.log(values)}")
print(f" the Sum is: {np.sum(values)}")
print(f" the Cumalative sum is: {np.cumsum(values)}")
```

## Task4

```python
data = np.arange(10,100,10).reshape((3,3))
print(f"The row wise sum is: {np.sum(data, axis=1)}")
print(f"The Column wise sum is: {np.sum(data, axis=0)}")
print(f"The Maximum is: {np.max(data)}")
print(f"The minimum is: {np.min(data)}")
print(f"The mean is: {np.mean(data)}")
```

## Task5

```python
marks = np.array([78,85,90,66,72,88,95,60])

print(f"Mean of marks is: {np.mean(marks)}")
print(f"Median of marks is: {np.median(marks)}")
print(f"Variance of marks is: {np.var(marks)}")
print(f"SD of marks is: {np.std(marks)}")
print(f"Min of marks is: {np.min(marks)}")
print(f"Max of marks is: {np.max(marks)}")
print(f"Range of marks is: {np.max(marks) - np.min(marks)}")
```

## Task6

```python
marks = np.sort(marks)

print(f"25th percentile: {np.percentile(marks, 25)}")
print(f"50th percentile: {np.percentile(marks, 50)}")
print(f"75th percentile: {np.percentile(marks, 75)}")
above_avg_students = marks > np.mean(marks)
print(f"No. of students above avg: {np.sum(above_avg_students)}")
```

## Task7

```python
sales = np.array([1200, 1500, 900, 2000, 1800, 1700, 1600])

print(f"Total weekly sales: {np.sum(sales)}")
print(f"Avg daily sale: {np.mean(sales)}")
print(f"Highest sale is {np.max(sales)} and lowest sale is {np.min(sales)}")
print(f"SD of sales data is: {np.std(sales)}")

above_avg_sales = sales > np.mean(sales)
index_for_above_avg_sales = np.where(above_avg_sales)
print(f"The days when sales is above avg are: {index_for_above_avg_sales}")
```
