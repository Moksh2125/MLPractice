```python
import joblib
import pandas as pd
```

```python
pipe = joblib.load("titanic_pipeline.pkl")
```

```python
sample = pd.DataFrame([{
    'pclass': 3,
    'sex': 'female',
    'age': 22,
    'sibsp': 1,
    'parch': 0,
    'fare': 7.25,
    'embarked': 'S'
}])

prediction = pipe.predict(sample)
prob = pipe.predict_proba(sample)

print("Prediction:", prediction[0])
print("Probability:", prob)
```
